```python
import pandas as pd

df= pd.read_csv("delivery_time.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df.distance_km, df.delivery_time_min)
plt.xlabel("Distance (KM)")
plt.ylabel("Delivery Time (min)")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(df[['distance_km']], df['delivery_time_min'])
```

```python
model.predict([[7],[34]])
```

```python
plt.scatter(df.distance_km,df.delivery_time_min,color='Black')
plt.plot(df.distance_km,model.predict(df[['distance_km']]))
plt.xlabel("Distance KM")
plt.ylabel("Delivery Time MIN")
plt.title("Best Fit line : time = m*distance+b")
plt.show()
```

```python
df2 = pd.read_csv('delivery_time_multi.csv')
df2.head()
```
